```python
import pandas as pd
customers   = pd.read_csv("../data/raw/olist_customers_dataset.csv")
orders      = pd.read_csv("../data/raw/olist_orders_dataset.csv")
order_items = pd.read_csv("../data/raw/olist_order_items_dataset.csv")
payments    = pd.read_csv("../data/raw/olist_order_payments_dataset.csv")
reviews     = pd.read_csv("../data/raw/olist_order_reviews_dataset.csv")
products    = pd.read_csv("../data/raw/olist_products_dataset.csv")
sellers     = pd.read_csv("../data/raw/olist_sellers_dataset.csv")
translation = pd.read_csv("../data/raw/product_category_name_translation.csv")
print("customers:", customers.shape)
print("orders:", orders.shape)
print("order_items:", order_items.shape)
print("payments:", payments.shape)
print("reviews:", reviews.shape)
print("products:", products.shape)
print("sellers:", sellers.shape)
print("translation:", translation.shape)

    
```

```python
orders["order_purchase_timestamp"]      = pd.to_datetime(orders["order_purchase_timestamp"])
orders["order_approved_at"]             = pd.to_datetime(orders["order_approved_at"])
orders["order_delivered_carrier_date"]  = pd.to_datetime(orders["order_delivered_carrier_date"])
orders["order_delivered_customer_date"] = pd.to_datetime(orders["order_delivered_customer_date"])
orders["order_estimated_delivery_date"] = pd.to_datetime(orders["order_estimated_delivery_date"])

orders.info()
```

```python
print("First order:", orders["order_purchase_timestamp"].min())
print("Last order :", orders["order_purchase_timestamp"].max())
```

```python
delivered = orders[orders["order_status"] == "delivered"].copy()

print("delivered orders:", len(delivered))
print("delivered but delivery date missing:", delivered["order_delivered_customer_date"].isnull().sum())
```

```python
delivered = delivered.dropna(subset=["order_delivered_customer_date"])

delivered["delivery_days"] = (delivered["order_delivered_customer_date"] - delivered["order_purchase_timestamp"]).dt.days

delivered["delay_days"] = (delivered["order_delivered_customer_date"].dt.normalize()
                           - delivered["order_estimated_delivery_date"].dt.normalize()).dt.days

delivered["is_late"] = (delivered["delay_days"] > 0).astype(int)

print("orders for delivery analysis:", len(delivered))
```

```python
delivered = delivered.dropna(subset=["order_delivered_customer_date"])

delivered["delivery_days"] = (delivered["order_delivered_customer_date"] - delivered["order_purchase_timestamp"]).dt.days

delivered["delay_days"] = (delivered["order_delivered_customer_date"].dt.normalize()
                           - delivered["order_estimated_delivery_date"].dt.normalize()).dt.days

delivered["is_late"] = (delivered["delay_days"] > 0).astype(int)

print("orders for delivery analysis:", len(delivered))
```

```python
print("order_items rows:", len(order_items), "| unique orders:", order_items["order_id"].nunique())
print("payments rows   :", len(payments),    "| unique orders:", payments["order_id"].nunique())
print("reviews rows    :", len(reviews),     "| unique orders:", reviews["order_id"].nunique())
print("duplicate review_id:", reviews["review_id"].duplicated().sum())
```

```python
reviews["review_answer_timestamp"] = pd.to_datetime(reviews["review_answer_timestamp"])

print("same review_id + same order_id duplicates:", reviews.duplicated(["review_id", "order_id"]).sum())

reviews_clean = reviews.sort_values("review_answer_timestamp").drop_duplicates(subset="order_id", keep="last")

print("reviews before:", len(reviews))
print("reviews after :", len(reviews_clean))
print("one review per order:", reviews_clean["order_id"].is_unique)
print(reviews_clean["review_score"].value_counts().sort_index())
```

```python
products_clean = products.merge(translation, on="product_category_name", how="left")

products_clean["category"] = (products_clean["product_category_name_english"]
                              .fillna(products_clean["product_category_name"])
                              .fillna("unknown"))

products_clean = products_clean[["product_id", "category"]]

print("products before:", len(products), "| after:", len(products_clean))
print("unknown category:", (products_clean["category"] == "unknown").sum())
print(products_clean["category"].value_counts().head(5))
```

```python
orders_clean = orders.merge(
    delivered[["order_id", "delivery_days", "delay_days", "is_late"]],
    on="order_id", how="left")

orders_clean["is_late"] = orders_clean["is_late"].astype("Int64")

print("orders before:", len(orders), "| after:", len(orders_clean))

orders_clean.to_csv("../data/processed/orders_clean.csv", index=False)
customers.to_csv("../data/processed/customers.csv", index=False)
order_items.to_csv("../data/processed/order_items.csv", index=False)
payments.to_csv("../data/processed/payments.csv", index=False)
reviews_clean.to_csv("../data/processed/reviews_clean.csv", index=False)
products_clean.to_csv("../data/processed/products_clean.csv", index=False)
sellers.to_csv("../data/processed/sellers.csv", index=False)
print("saved")
```
